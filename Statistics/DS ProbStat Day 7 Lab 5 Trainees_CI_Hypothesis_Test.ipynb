## Confidence Interval (CI)
### Within how many standard deviations the estimated sample mean should be located for CIs 90%, 92%, 95%, 98%, 99%.   

```python
from scipy.stats import norm

confidence_levels = [0.90, 0.92, 0.95, 0.98, 0.99]

for cl in confidence_levels:
    z = norm.ppf(1 - (1-cl)/2)
    print(f"{(cl*100)}% CI: within {z} std") 

#Probability of the outside the confidence interval cl=1-0.95 = 0.05
#then we will get the tails since that they are 2 
#0.05/2 = 0.025
#so in order to get the upper bound commulative probab = 1-0.025
```

### Calculate the expected mean salary with 95% CI for each team for the employees data as follow:
##### Using z-value.
##### Using t-value.
##### Compare the results obtaind from the above calculation and write a conclusion.
##### Perform  sns.barplot() and check out the displayed confidence intervals. Compare with your results.

```python
# teams = df.groupby('Team')['Salary']
# print(teams.mean(),teams.count(),teams.std())
#z-score when the std of the population is known 
#for pop > 30 uses normal distribution mean = 0 and std = 1
#we use the tscore when we dont have the std of the population
teams = df.groupby('Team')['Salary'].agg(['mean', 'count', 'std'])
print(teams)
```

```python
from scipy.stats import t

z_value = norm.ppf(1 - (1-0.95)/2) 
#CI Lower : mean - z value * sample std / root sample size
teams['CI_z_lower'] = teams['mean'] - z_value * (teams['std'] / np.sqrt(teams['count']))

#CI Upper : mean + z value * sample std / root sample size
teams['CI_z_upper'] = teams['mean'] + z_value * (teams['std'] / np.sqrt(teams['count']))

teams['CI_t_lower'] = teams.apply(lambda row: row['mean'] - t.ppf(1 - (1-0.95)/2, df=row['count']-1) * (row['std'] / np.sqrt(row['count'])), axis=1)
teams['CI_t_upper'] = teams.apply(lambda row: row['mean'] + t.ppf(1 - (1-0.95)/2, df=row['count']-1) * (row['std'] / np.sqrt(row['count'])), axis=1)

pd.set_option('display.float_format', lambda x: '%.2f' % x)
print(teams)
```
