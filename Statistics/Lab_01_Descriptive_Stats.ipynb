## Titanic Data Analysis
[Titanic Data Set from Kaggle](https://www.kaggle.com/c/titanic).
##### Features:
- survival - Survival (0 = No; 1 = Yes)
- pclass - Passenger Class (1 = 1st; 2 = 2nd; 3 = 3rd)
- name - Name
- sex - Sex
- age - Age
- sibsp - Number of Siblings/Spouses Aboard
- parch - Number of Parents/Children Aboard
- ticket - Ticket Number
- fare - Passenger Fare
- cabin - Cabin
- embarked - Port of Embarkation (C = Cherbourg; Q = Queenstown; S = Southampton)

### 17. Display the distribution of the ticket fare (Fare) with mean, mode, and median

```python
#your insights goes here
#Mean > Median > Mode 
#Data is right skewed
```

```python
fare = df['Fare'].dropna()
mean_fare = fare.mean()
median_fare = fare.median()
mode_fare = fare.mode()[0]

plt.hist(fare, bins=50, edgecolor='black', alpha=0.6, color='skyblue')
plt.axvline(mean_fare, color='red', linewidth=2, label=f'Mean: {mean_fare:.2f}')
plt.axvline(median_fare, color='green', linewidth=2, label=f'Median: {median_fare:.2f}')
plt.axvline(mode_fare, color='orange', linewidth=2, label=f'Mode: {mode_fare:.2f}')

plt.xlabel('Fare')
plt.ylabel('Count')
plt.legend()
plt.show()
```

### 18. Display the distribution of the ticket fare (Fare) with histplot and boxplot

```python
fare = df['Fare'].dropna()
mean_fare = fare.mean()
median_fare = fare.median()
mode_fare = fare.mode()[0]

plt.figure(figsize=(10,6))

plt.subplot(2,1,1)
sns.boxplot(data=df, x='Fare')

plt.subplot(2,1,2)
plt.hist(fare, bins=30, edgecolor='black', alpha=0.6, color='skyblue')
plt.axvline(mean_fare, color='red', linewidth=2, label=f'Mean: {mean_fare:.2f}')
plt.axvline(median_fare, color='green', linewidth=2, label=f'Median: {median_fare:.2f}')
plt.axvline(mode_fare, color='orange', linewidth=2, label=f'Mode: {mode_fare:.2f}')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.legend()

plt.tight_layout()
plt.show()
```
