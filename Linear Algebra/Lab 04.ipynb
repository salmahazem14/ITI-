#### Generate three random vectors in $R^3$ by

- If the answer is not 3, then generate a new random set of vectors and calculate the rank. Repeat until the rank is 3. Now use these vectors in the following calculations.

```python
#since the rank is 3 , then the 3 vectors are linearly independent 
#Full rank since rows = cols = rank

u1 = np.array([[4], [9], [6]])
u2 = np.array([[6], [8], [8]])
u3 = np.array([[0], [9], [6]])
U = np.block([u1, u2, u3])

print("-------u1--------")
print(u1)
print("--------u2---------")
print(u2)
print("--------u3---------")
print(u3)
print("-----------------------")

rank = np.linalg.matrix_rank(U)

print("Rank :")
print(rank)
```

#### (b) Now use the vectors u1, u2, u3 to obtain an orthogonal basis for R3, following the Gram-Schmidt algorithm. Set `v1 = u1`. Obtain v2 by removing the component of u2 in the direction v1:


\\[v_2 = u_2 - \frac{v_1 \cdot u_2}{v_1 \cdot v_1}v_1\\]

- Calculate the dot product to check that the vectors v1 and v2 are mutually orthogonal (within a negligible numerical error). Also add v2 to your graphics window as a dashed-dotted green line.

```python
#type your code here
```

```python
#What we are trying to do here is : 
#we found out that u1,u2,u3 are not perpendicular 
#We use gram-schmidt to generate vectors that make them mutually orthogonal 
#Mutually orthogonal ==> having dot product of 0 
```

```python
v1 = u1

#v1.T @ u2 ===> scalar value showing how much of u2 points 
# in v1's direction
#v1.T @ v1 --> dot product of v1 with it self 

#Makes v2 perpendicular to v1 
v2 = u2 - (v1.T @ u2 / (v1.T @ v1)) * v1 #multiplying by v1 at the end maintains direction

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot3D(r * v1[0], r * v1[1], r * v1[2], 'r-.', label='v1')  
ax.plot3D(r * v2[0], r * v2[1], r * v2[2], 'g-.', label='v2')

ax.set_xlim([0,10])
ax.set_ylim([0,10])
ax.set_zlim([0,10])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.legend()
plt.show()
```

#### (c) The last step in the Gram-Schmidt algorithm is to rescale the vectors v1, v2, v3 to obtain an orthonormal basis for R3:

- `w1 = v1 / np.linalg.norm(v1)`
- `w2 = v2 / np.linalg.norm(v2)`
- `w3 = v3 / np.linalg.norm(v3)`

#### Define the matrix Q = [w1, w2, w3] and give written answers to the following questions.



- What is the inverse matrix Q Now.


Q.T @ Q = I then Q ^ -1 = Q.T 
so the inverse of matrix Q is its transpose 

```python
#type your code here:
```

```python
#Orthonormal vectors 
#started by rescaling 
w1 = v1 / np.linalg.norm(v1)
w2 = v2 / np.linalg.norm(v2)
w3 = v3 / np.linalg.norm(v3)

print("------w1------")
print(w1)
print("------w2------")
print(w2)
print("------w3------")
print(w3)

#Matrix out of rescaling
Q = np.hstack([w1, w2, w3]) 

print("------Q------")
print(Q)

Q_inv = Q.T  
#Resulted in identity 
print("------Q.T @ Q ------")
print(Q.T @ Q) 

#Transpose of Q
print("------Q inverse ------")
print(Q_inv)

```
